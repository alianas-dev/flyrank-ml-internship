## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision: which pages a content reviewer should look at first, out of thousands, when they only have time to review a handful each week.
Who acts, and how: a content editor/SEO reviewer opens the top of a ranked "review queue" and decides whether to refresh, expand, or leave a page, instead of guessing or going alphabetically.
Cost of a wrong call: if the system flags a healthy page as "at risk," the reviewer wastes time on a page that didn't need help (low cost, just wasted effort). If the system misses a genuinely declining page, that page keeps losing traffic unnoticed until someone catches it later (higher cost lost visibility compounds over time). Because a missed decline is worse than a wasted review, the system should lean toward flagging borderline cases rather than staying silent on them.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how many pages are currently declining
declining_rate = (df["trend_direction"].str.lower() == "down").mean()
print(f"1) {declining_rate:.1%} of pages are currently in a 'down' trend — a real, sizeable pool to prioritize.")

# Number 2: model vs rule lift — verified numbers from the Week 1 pipeline run
# (outputs/model_results.json / outputs/model_report.md, confirmed in the lane guide)
base = 0.240
rf = 0.740
print(f"2) A learned model beat a hand-written rule by {rf/base:.1f}x on Precision@50 ({base:.3f} -> {rf:.3f}) — real learnable signal exists.")

# Number 3: how many pages get real search visibility, worth prioritizing
visible_pages = (df["impressions_90d"] >= 500).sum()
print(f"3) {visible_pages:,} pages have 500+ impressions in 90 days — enough traffic to matter if they're declining.")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I can claim: this system produces observed, directional signals,  e.g. "this page shows patterns similar to other pages that later declined" or "this page is a reasonable candidate to review first." All results are decision-support: they help a human prioritize, not replace judgment.
What I cannot claim: that a flagged page is guaranteed to decline, that refreshing a page causes recovery (that needs a real experiment, not this data), or that any of this reveals how Google's ranking algorithm actually works. Correlation and pattern-matching are not proof of causation.
